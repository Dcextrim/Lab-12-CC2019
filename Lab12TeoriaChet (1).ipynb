### Ejercicio 1: Ordenar una lista de tuplas

Este ejercicio define una lista de tuplas `d` con información de productos y la ordena basándose en el segundo elemento de cada tupla (el modelo).

```python
%%writefile Ejercicio1.hs
import Data.List (sortBy)
import Data.Ord (comparing)

-- Lista
d :: [
     (String, Int, String)
     ]

d = [
     ("Nokia", 216, "Black"),
     ("Apple", 2, "Silver"),
     ("Huawei", 50, "Gold"),
     ("Samsung", 7, "Blue")
    ]

-- Ordenar por modelo
ordenar :: [(String, Int, String)] -> [(String, Int, String)]
ordenar = sortBy (comparing (\(_, modelo, _) -> modelo))

main = print (ordenar d)
```

```python
import os
import subprocess
import ast

program_name = 'Ejercicio1.hs'
executable_name = 'Ejercicio1'

print(f"Compilando {program_name}...")
compile_result = subprocess.run(['ghc', program_name], capture_output=True, text=True)
print("GHC stdout:\n", compile_result.stdout)
print("GHC stderr:\n", compile_result.stderr)
print(f"GHC exit code: {compile_result.returncode}")

if compile_result.returncode == 0:
    print(f"\nCompilación exitosa. Ejecutando {executable_name}...")
    run_result = subprocess.run([f'./{executable_name}'], capture_output=True, text=True)
    print("Executable stdout:\n")

    try:
        parsed_output = ast.literal_eval(run_result.stdout.strip())
        for item in parsed_output:
            print(item)
    except (ValueError, SyntaxError) as e:
        print(f"Error al parsear la salida: {e}")
        print("Raw Executable stdout:\n", run_result.stdout)

    print("Executable stderr:\n", run_result.stderr)
    print(f"Executable exit code: {run_result.returncode}")
else:
    print("\nLa compilación falló para Ejercicio1. Por favor, revisa el GHC stderr para más detalles.")
```

### Ejercicio 3: Transponer una matriz

Este ejercicio define una función para transponer una matriz (lista de listas).

```python
%%writefile Ejercicio3.hs
transponerM :: [[a]] -> [[a]]
transponerM = foldr (zipWith (:)) (repeat [])

main = print (transponerM [[1,2,3,1],
                           [4,5,6,0],
                           [7,8,9,-1]]
                           )
```

```python
import os
import subprocess
import ast

program_name = 'Ejercicio3.hs'
executable_name = 'Ejercicio3'

print(f"Compilando {program_name}...")
compile_result = subprocess.run(['ghc', program_name], capture_output=True, text=True)
print("GHC stdout:\n", compile_result.stdout)
print("GHC stderr:\n", compile_result.stderr)
print(f"GHC exit code: {compile_result.returncode}")

if compile_result.returncode == 0:
    print(f"\nCompilación exitosa. Ejecutando {executable_name}...")
    run_result = subprocess.run([f'./{executable_name}'], capture_output=True, text=True)
    print("Executable stdout:\n")

    try:
        parsed_output = ast.literal_eval(run_result.stdout.strip())
        for row in parsed_output:
            print(row)
    except (ValueError, SyntaxError) as e:
        print(f"Error al parsear la salida: {e}")
        print("Raw Executable stdout:\n", run_result.stdout)

    print("Executable stderr:\n", run_result.stderr)
    print(f"Executable exit code: {run_result.returncode}")
else:
    print("\nLa compilación falló para Ejercicio3. Por favor, revisa el GHC stderr para más detalles.")
```

### Ejercicio 4: Eliminar elementos de una lista

Este ejercicio define una función que elimina todos los elementos de una segunda lista que estén presentes en la primera lista.

```python
%%writefile Ejercicio4.hs
eliEl :: Eq a => [a] -> [a] -> [a]
eliEl xs borrar = filter (\x -> not (x `elem` borrar)) xs

main = print (eliEl ["rojo","verde","azul","amarillo","gris","blanco","negro"]
                       ["amarillo","café","blanco"])
```

```python
import os
import subprocess
import ast

program_name = 'Ejercicio4.hs'
executable_name = 'Ejercicio4'

print(f"Compilando {program_name}...")
compile_result = subprocess.run(['ghc', program_name], capture_output=True, text=True)
print("GHC stdout:\n", compile_result.stdout)
print("GHC stderr:\n", compile_result.stderr)
print(f"GHC exit code: {compile_result.returncode}")

if compile_result.returncode == 0:
    print(f"\nCompilación exitosa. Ejecutando {executable_name}...")
    run_result = subprocess.run([f'./{executable_name}'], capture_output=True, text=True)
    print("Executable stdout:\n")

    try:
        parsed_output = ast.literal_eval(run_result.stdout.strip())
        for item in parsed_output:
            print(item)
    except (ValueError, SyntaxError) as e:
        print(f"Error al parsear la salida: {e}")
        print("Raw Executable stdout:\n", run_result.stdout)

    print("Executable stderr:\n", run_result.stderr)
    print(f"Executable exit code: {run_result.returncode}")
else:
    print("\nLa compilación falló para Ejercicio4. Por favor, revisa el GHC stderr para más detalles.")
```
